## DC bus and grid inverter tracker

### What this plot illustrates

Electrical balance: DC bus voltage and inverter/grid interface tracking a power command $P_\mathrm{grid}(t)$.

### Governing equations

$P_\mathrm{dc}=V_\mathrm{bus} I_\mathrm{bus}$, $P_\mathrm{ac}\approx \eta_\mathrm{inv} P_\mathrm{dc}$ (first-order tracker, not switching-resolved).


```python
plot_frame = pd.DataFrame(
    {
        "t": (res["t"]),
        "Vdc": (res["y"][0]),
        "Pinv": (res["y"][3]),
    }
)
plot_long = plot_frame.melt(id_vars="t", var_name="quantity", value_name="value")
fig, ax = plt.subplots()
sns.lineplot(data=plot_long, x="t", y="value", hue="quantity", ax=ax, linewidth=2.0)
ax.set_xlabel("t (s)")
ax.set_ylabel("V / W")
ax.set_title("DC bus and inverter tracker")
ax.text(0.0, -0.22, "IA-DC-001 IA-INV-001", transform=ax.transAxes, fontsize=8, va="top")
fig.tight_layout()
fig.subplots_adjust(bottom=0.22)
```
